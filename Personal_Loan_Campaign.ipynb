```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.drop('Personal_Loan', axis=1),
    df['Personal_Loan'],
    test_size=0.2,
    random_state=42,
    stratify=df['Personal_Loan']
)
```

```python
y_train.value_counts(normalize=True)
y_test.value_counts(normalize=True)
```

```python
m1 = DecisionTreeClassifier(
    class_weight='balanced',
    random_state=42
)
```

```python
m1.fit(X_train, y_train)
```

```python
y_pred = m1.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state=42)
```

```python
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)
```

```python
m2 = DecisionTreeClassifier(random_state=42)
```

```python
m2.fit(X_train_smote, y_train_smote)
```

```python
y_pred_m2 = m2.predict(X_test)
```

```python
print("Classification Report:")
print(classification_report(y_test, y_pred_m2))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_m2))
```

```python
print("Model 1 train accuracy:", m1.score(X_train, y_train))
print("Model 1 test accuracy:", m1.score(X_test, y_test))
```

```python

print("Model 2 train accuracy:", m2.score(X_train_smote, y_train_smote))
print("Model 2 test accuracy:", m2.score(X_test, y_test))
```

```python
m1.get_depth()
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': [5, 8, 11, 14, 17, None],
    'min_samples_leaf': [20, 40, 60, 80, 100],
    'min_samples_split': [40, 80, 120, 160, 200]
}
```

```python
# Model 1: class_weight version — search on original data
grid_search_1 = GridSearchCV(
    estimator=DecisionTreeClassifier(class_weight='balanced', random_state=42),
    param_grid=param_grid,
    scoring='f1',
    cv=5,
    n_jobs=-1
)
grid_search_1.fit(X_train, y_train)

print("Best params (Model 1):", grid_search_1.best_params_)
print("Best CV F1 score (Model 1):", grid_search_1.best_score_)

best_m1 = grid_search_1.best_estimator_
y_pred_best_m1 = best_m1.predict(X_test)

print("\nModel 1 (tuned) Classification Report:")
print(classification_report(y_test, y_pred_best_m1))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_best_m1))
```

```python
# Model 2: SMOTE version — search on SMOTE-balanced data
grid_search_2 = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    scoring='f1',
    cv=5,
    n_jobs=-1
)
grid_search_2.fit(X_train_smote, y_train_smote)

print("\nBest params (Model 2):", grid_search_2.best_params_)
print("Best CV F1 score (Model 2):", grid_search_2.best_score_)

best_m2 = grid_search_2.best_estimator_
y_pred_best_m2 = best_m2.predict(X_test)

print("\nModel 2 (tuned) Classification Report:")
print(classification_report(y_test, y_pred_best_m2))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_best_m2))


```
